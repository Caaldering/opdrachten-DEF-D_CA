## *Opdracht 6*: Meten en fitten, Leerdoel

```python
lux = [950, 650, 430, 145, 100, 65]
output = [920,  870, 820, 600, 480, 310]

#onzekerheid +/- 10 lux en 15 output

```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.

import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

plt.figure(figsize=(10, 5))
plt.plot(lux, output, 'o', label='Metingen')
plt.xlabel('Lichtintensiteit (lux)')
plt.ylabel('Output')
plt.title('Kalibratie van de lichtsensor')
plt.grid()
plt.legend()
plt.show()

def fitfunctie(x, a, b, c):
    return a * np.log(b * x) + c


popt, pcov = curve_fit(fitfunctie, lux, output)
a_fit = popt[0]
b_fit = popt[1]
c_fit = popt[2]

print(f'Fitparameter a: {a_fit}')
print(f'Fitparameter b: {b_fit}')
print(f'Fitparameter c: {c_fit}')

x_fit = np.linspace(min(lux), max(lux), 100)
y_fit = fitfunctie(x_fit, a_fit, b_fit, c_fit)
plt.figure(figsize=(10, 5))
plt.plot(lux, output, 'o', label='Metingen')
plt.plot(x_fit, y_fit, '-', label='Fitfunctie')
plt.errorbar(lux, output, xerr=10, yerr=15, fmt='o', label='Metingen met onzekerheid')
plt.xlabel('Lichtintensiteit (lux)')
plt.ylabel('Output')
plt.title('Kalibratie van de lichtsensor met fit')
plt.grid()
plt.legend()
```

## *Opdracht 7*: Iteratie nodig?

Nee, kalibaratie is goed gelukt. Er is een logaritmisch verband tussen de lux en de output.
